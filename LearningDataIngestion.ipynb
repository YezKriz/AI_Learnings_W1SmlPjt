```python
long_text = ''' Python is a programming language that lets you work quickly and integrate systems more effectively. 
Streamlit is used to build web apps for machine learning and data science. Langchain is a framework for building applications with LLMs. 
RAG combines the power of LLMs with external knowledge sources to provide more accurate and relevant responses. Embeddings are numerical representations of text that capture its semantic meaning.
Vector databases are used to store and search embeddings efficiently. Retrivers find the most relevant documents from a knowledge''' 

print("Total number of characters in the text:", len(long_text))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=80, # each chunk is 80 characters long
    chunk_overlap=10 # each chunk overlaps with the previous one by 10 characters
)

chunks = splitter.split_text(long_text)

print("Total number of chunks created:", len(chunks))
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}")
    print(chunk)
    
```
